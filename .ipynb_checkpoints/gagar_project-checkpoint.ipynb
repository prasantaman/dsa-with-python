```python
import pandas as pd
import re

# Excel file read
df = pd.read_excel("input.xlsx")

def convert_item(row):
    item = str(row["Items"])
    count = float(row["Quantity"])

    # Pattern for weight/volume
    pattern = r'(\d+(?:\.\d+)?)\s*(Gram|gm|g|Kg|ML|Ltr|Litre|Pkt)$'

    match = re.search(pattern, item, re.IGNORECASE)

    if match:
        amount = float(match.group(1))
        unit = match.group(2)

        # Item name without quantity part
        new_item = re.sub(pattern, '', item, flags=re.IGNORECASE).strip()

        # Multiply by packet count
        total_amount = amount * count

        # Convert grams to kg
        if unit.lower() in ['gram', 'gm']:
            quantity = round(total_amount / 1000, 3)
            unit_name = "kg"

        # Convert ml to litre
        elif unit.lower() == "ml":
            quantity = round(total_amount / 1000, 3)
            unit_name = "litre"

        else:
            quantity = total_amount
            unit_name = unit

        return pd.Series([new_item, quantity, unit_name])

    else:
        # No size mentioned
        return pd.Series([item, count, "Nos"])

# Create new columns
df[["New_Item", "Quantity", "Unit"]] = df.apply(convert_item, axis=1)

# Save output
df.to_excel("output.xlsx", index=False)

print("Done")
```
